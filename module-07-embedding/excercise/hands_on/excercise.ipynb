```python
documents = [
    "Artificial intelligence is changing modern technology.",  # Technology
    "Machine learning helps computers learn from data.",  # Technology
    "Neural networks are widely used in artificial intelligence.",  # Technology
    "New computer processors are becoming faster.",  # Technology
    "The football team won the championship.",  # sport
    "The coach trained the team for the tournament.",  # sport
    "The player scored three goals in the match.",  # sport
    "Fresh vegetables make a healthy salad.",  # Food
    "Pizza is a popular Italian food.",  # Food
    "The chef prepared a delicious pasta dish.",  # Food
]
```

```python
class RetriveDocument:
    def __init__(self):
        self.model = SentenceTransformer("all-MiniLM-L6-v2")
        self.tokenizer = tiktoken.get_encoding("cl100k_base")

    def build_search_index(self, documents: list[str]) -> list[tuple[str, np.ndarray]]:
        chunks = [chunk for doc in documents for chunk in self._chunk_text(doc)]
        embedding = self._build_embeddings(chunks)

        return list(zip(chunks, embedding))

    def semantic_search(self, query: str, index: list[tuple[str, np.ndarray]], top_k: int = 3):
        query_embedding = self._build_embeddings([query])
        scored = [
            (
                chunk,
                cosine_similarity(query_embedding, chunk_embedding.reshape(1, -1))[0][0],
            )
            for chunk, chunk_embedding in index
        ]

        scored.sort(key=lambda x: x[1], reverse=True)

        return scored[:top_k]

    def _chunk_text(self, text: str, chunk_size_tokens: int = 500, overlaps_token: int = 50):
        tokens = self.tokenizer.encode(text)
        chunks = []
        start = 0
        while start < len(tokens):
            end = min(start + chunk_size_tokens, len(tokens))
            chunks.append(self.tokenizer.decode(tokens[start:end]))
            start += chunk_size_tokens - overlaps_token
        return chunks

    def _build_embeddings(self, texts: list[str]) -> np.ndarray:
        return self.model.encode(texts)
```

```python
retrive_document = RetriveDocument()
```

```python
index = retrive_document.build_search_index(documents)
```

```python
retrivals = retrive_document.semantic_search(query="Who is changing modern technology", index=index)
```

## task 6

```python
import numpy as np
from sentence_transformers import SentenceTransformer

e5 = SentenceTransformer("intfloat/e5-base-v2")
minilm = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

documents = [
    "Python is a programming language commonly used for data science and machine learning.",
    "The Eiffel Tower is a famous landmark located in Paris, France.",
    "Neural networks are machine learning models inspired by the human brain.",
    "The Pacific Ocean is the largest ocean on Earth.",
    "Pandas is a Python library used for data manipulation and analysis.",
    "Transformers are neural network architectures widely used in natural language processing.",
    "Mount Everest is the highest mountain above sea level.",
    "SQL is a language used to query and manipulate relational databases.",
]

queries = [
    "What Python library is useful for manipulating datasets?",
    "Which architecture is widely used for NLP?",
    "What is the highest mountain in the world?",
    "Which ocean is the biggest?",
]


# Ground-truth relevant document for each query
relevant = {
    0: 4,  # Pandas
    1: 5,  # Transformers
    2: 6,  # Everest
    3: 3,  # Pacific Ocean
}


def retrieve(model, queries, documents, asymmetric=False, top_k=3):
    if asymmetric:
        # E5 requires special prefixes.
        query_texts = [f"query: {q}" for q in queries]
        document_texts = [f"passage: {d}" for d in documents]
    else:
        # Symmetric/general model: same treatment on both sides.
        query_texts = queries
        document_texts = documents

    query_embeddings = model.encode(query_texts, normalize_embeddings=True)

    document_embeddings = model.encode(document_texts, normalize_embeddings=True)

    # Because embeddings are normalized, dot product = cosine similarity.
    scores = query_embeddings @ document_embeddings.T

    results = []

    for query_idx, query_scores in enumerate(scores):
        ranked = np.argsort(query_scores)[::-1][:top_k]

        results.append(
            {
                "query": queries[query_idx],
                "documents": [
                    {
                        "rank": rank + 1,
                        "index": int(doc_idx),
                        "score": float(query_scores[doc_idx]),
                        "text": documents[doc_idx],
                    }
                    for rank, doc_idx in enumerate(ranked)
                ],
            }
        )

    return results


def recall_at_k(results, relevant, k):
    correct = 0

    for query_idx, result in enumerate(results):
        retrieved = [item["index"] for item in result["documents"][:k]]

        if relevant[query_idx] in retrieved:
            correct += 1

    return correct / len(results)


def mrr_at_k(results, relevant, k):
    reciprocal_ranks = []

    for query_idx, result in enumerate(results):
        retrieved = [item["index"] for item in result["documents"][:k]]

        target = relevant[query_idx]

        if target in retrieved:
            rank = retrieved.index(target) + 1
            reciprocal_ranks.append(1 / rank)
        else:
            reciprocal_ranks.append(0)

    return sum(reciprocal_ranks) / len(reciprocal_ranks)


e5_results = retrieve(e5, queries, documents, asymmetric=True, top_k=3)

minilm_results = retrieve(minilm, queries, documents, asymmetric=False, top_k=3)


def print_results(name, results):
    print("\n" + "=" * 70)
    print(name)
    print("=" * 70)

    for i, result in enumerate(results):
        print(f"\nQuery: {result['query']}")

        for doc in result["documents"]:
            marker = " <-- RELEVANT" if doc["index"] == relevant[i] else ""

            print(f"{doc['rank']}. {doc['score']:.4f} | {doc['text']}{marker}")


print_results("E5 (Asymmetric Retrieval)", e5_results)
print_results("MiniLM (Symmetric)", minilm_results)

print("\n" + "=" * 70)
print("METRICS")
print("=" * 70)

print(f"E5     Recall@3: {recall_at_k(e5_results, relevant, 3):.3f}")

print(f"E5     MRR@3:    {mrr_at_k(e5_results, relevant, 3):.3f}")

print(f"MiniLM Recall@3: {recall_at_k(minilm_results, relevant, 3):.3f}")

print(f"MiniLM MRR@3:    {mrr_at_k(minilm_results, relevant, 3):.3f}")
```
