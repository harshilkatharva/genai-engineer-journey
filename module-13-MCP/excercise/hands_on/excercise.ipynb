```python
mcp = MCPServer("Company Knowledge Server")
```

```python
# Bad: overly broad tool
db = ""  # real database connectivities


@mcp.tool()
def database_query(sql: str) -> list:
    """Execute arbitrary SQL in the sandbox database."""
    return db.execute(sql)
```

```python
# Good: least-privilege tool

order_service = ""


@mcp.tool()
def get_order_status(order_id: str) -> dict:
    """Read the status of one specific order."""
    if not re.fullmatch(r"ORD-\d{8}", order_id):
        raise ValueError("Invalid order ID format")

    order = order_service.get(order_id)

    return {"status": order.status, "estimated_delivery": order.eta}
```
